```python
mbfcs = pd.read_csv('data/mbfc_source_data.csv')
def replace_nans(text_list, replacement=""):
    return ["" if str(x).strip().lower() == "nan" or x != x else x for x in text_list]

def target(df, mbfcs):
    df = df.drop_duplicates(subset="text", keep="first").reset_index(drop=True)
    df['factual_reporting'] = ''  # Initialize column as object type
    df['factual_reporting'] = df['factual_reporting'].astype('object')
    # Normalize mbfcs keys for matching
    mbfcs['normalized_display_name'] = mbfcs['display_name'].str.lower().str.replace(" ", "", regex=False)
    mbfcs['normalized_website'] = mbfcs['website'].apply(lambda x: urlparse(x).hostname if pd.notnull(x) else '').str.lower()

    for i in df.index:
        
        target_clean = str(df.loc[i, 'target']).lower().replace(" ", "")
        domain_clean = str(df.loc[i, 'domain']).lower().replace(" ", "")

        match = mbfcs[mbfcs['normalized_display_name'] == target_clean]
        if not match.empty:
            df.at[i, 'factual_reporting'] = match.iloc[0]['factual_reporting']
            continue

        match = mbfcs[mbfcs['normalized_website'] == domain_clean]
        if not match.empty:
            df.at[i, 'factual_reporting'] = match.iloc[0]['factual_reporting']
        else:
            df.at[i, 'factual_reporting'] = None
    df = df.dropna(axis=0, subset=['text'])
    return df


d_20171031_NY = target(d_20171031_NY, mbfcs)
d_20170812_VA = target(d_20170812_VA, mbfcs)
d_20160612_FL = target(d_20160612_FL, mbfcs)
d_20250602_CO = target(d_20250602_CO, mbfcs)
d_20140524_CA = target(d_20140523_CA, mbfcs)

result_series = set(d_20140524_CA.loc[d_20140524_CA['factual_reporting'].isnull(), 'domain']) | \
                set(d_20171031_NY.loc[d_20171031_NY['factual_reporting'].isnull(), 'domain'])| \
                set(d_20170812_VA.loc[d_20170812_VA['factual_reporting'].isnull(), 'domain'])| \
                set(d_20160612_FL.loc[d_20160612_FL['factual_reporting'].isnull(), 'domain'])| \
                set(d_20250602_CO.loc[d_20250602_CO['factual_reporting'].isnull(), 'domain'])
```

```python
def make_matrix(topics):
    ctfidf = topics.c_tf_idf_.toarray()
    topic_indices = list(topics.topic_labels_.keys())
    vectors = ctfidf[topic_indices]
    umap_model = umap.UMAP(n_neighbors=2, n_components=2, metric="hellinger", random_state=42)
    scaler = MinMaxScaler()
    scaled = scaler.fit_transform(vectors)
    coords = umap_model.fit_transform(scaled)

    distance_matrix = pd.DataFrame(cdist(coords, coords), 
                                index=[f"Topic {i}" for i in range(len(coords))], 
                                columns=[f"Topic {i}" for i in range(len(coords))])

    return coords, distance_matrix
```

```python
def merge_topics_by_global_std_indices(distance_df, threshold_std=2):
    # Convert to numpy matrix
    mat = distance_df.values

    # Get all off-diagonal upper triangle values
    off_diag = mat[np.triu_indices_from(mat, k=1)]

    # Compute global avg and std
    avg = np.mean(off_diag)
    std = np.std(off_diag)
    cutoff = avg - threshold_std * std

    print(f"Global avg: {avg:.4f}, std: {std:.4f}, cutoff: {cutoff:.4f}")

    visited = set()
    clusters = []

    for i, topic in enumerate(distance_df.index):
        if topic in visited:
            continue

        # Find indices of topics whose distance is below cutoff
        distances = distance_df.loc[topic]
        close_topics = distances[distances < cutoff].index.tolist()

        group = sorted(set([topic] + close_topics))
        if len(group) > 1:
            clusters.append([int(t.split()[-1]) if isinstance(t, str) and t.startswith("Topic") else int(t) for t in group])
            visited.update(group)

    return clusters
```

```python
def calc_avg_per_column(df):
    # Extract the upper triangle of the matrix without the diagonal
    upper_triangle_values = df.where(np.triu(np.ones(df.shape), k=1).astype(bool)).stack()

    df_no_diag = df.mask(np.eye(len(df), dtype=bool))

    # Calculate inter-topic average for each column (excluding diagonal)
    intertopic_avg_per_column = df_no_diag.mean(axis=0)
    intertopic_std_per_column = df_no_diag.std(axis=0)
    print(f"avg:{intertopic_avg_per_column} stdev: {intertopic_std_per_column}")

    within_stdev = {}
    for topic, avg in intertopic_avg_per_column.items():
        std = intertopic_std_per_column[topic]
        lower = avg - std
        upper = avg + std
        similar_topics = intertopic_avg_per_column[(intertopic_avg_per_column <= lower)].index.tolist()
        print('topic:',topic,'similar:', similar_topics)
        #similar_topics.remove(topic)  # exclude self
        within_stdev[topic] = similar_topics

    return within_stdev
```

```python
def sort_ratio(df):
    # Ensure class_ratio column is a dictionary
    df = df.drop_duplicates(subset="text", keep="first").reset_index(drop=True)

    df["class_ratio"] = df["class_ratio"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else x)

    # Find the highest class ID present in any row
    max_class = max(
        (max(ratio.keys()) for ratio in df["class_ratio"] if isinstance(ratio, dict) and ratio),
        default=0
    )

    # Create a new column for that max class's ratio
    df["max_class_ratio"] = df["class_ratio"].apply(lambda r: r.get(max_class, 0.0))

    # Sort by that column descending
    df_sorted = df.sort_values(by="max_class_ratio", ascending=False)

    return df_sorted
```

```python
def pair_closest_topics(df):
    mat = df.values

    # Get all off-diagonal upper triangle values
    off_diag = mat[np.triu_indices_from(mat, k=1)]
    avg = np.mean(off_diag)
    std = np.std(off_diag)  
# Drop column and row labeled 'Topic -1' if it exists
    if 'Topic -1' in df.columns:
        df = df.drop(columns='Topic -1', errors='ignore')
    if 'Topic -1' in df.index:
        df = df.drop(index='Topic -1', errors='ignore')

    pairs = {}
    for topic in df.columns:
        distances = df[topic].copy()
        distances[topic] = np.inf  # Ignore self-distance
        closest_topic = distances.idxmin()
        pairs[topic] = closest_topic
    return pairs, avg,std

# Get closest topic pairs
closest_pairs,avg,std = pair_closest_topics(tm_20160612_FL_dist_df)
closest_pairs_df = pd.DataFrame(list(closest_pairs.items()), columns=["Topic", "Closest_Topic"])
```

```python
def pair_closest_topics_filtered(df):
    # Drop column and row labeled 'Topic -1' if it exists
    if 'Topic -1' in df.columns:
        df = df.drop(columns='Topic -1', errors='ignore')
    if 'Topic -1' in df.index:
        df = df.drop(index='Topic -1', errors='ignore')

    # Compute overall average and standard deviation of off-diagonal values
    mat = df.values
    off_diag = mat[np.triu_indices_from(mat, k=1)]
    avg = np.mean(off_diag)
    std = np.std(off_diag)
    cutoff = avg - std

    # Pair topics only if their closest distance is below the cutoff
    pairs = {}
    for topic in df.columns:
        distances = df[topic].copy()
        distances[topic] = np.inf  # Ignore self-distance
        min_value = distances.min()
        closest_topic = distances.idxmin()
        pairs[topic] = closest_topic if min_value < cutoff else None

    return pairs, avg, std

# Apply the updated function
filtered_pairs, global_avg, global_std = pair_closest_topics_filtered(tm_20250602_CO_dist_df)
filtered_pairs_df = pd.DataFrame(list(filtered_pairs.items()), columns=["Topic", "Closest_Topic"])

```
