```python
def split_df(df: pd.DataFrame, seed: int = 42):
    # Shuffle once
    df_shuffled = df.sample(frac=1, random_state=seed).reset_index(drop=True)

    n = len(df_shuffled)

    # Compute exact split indices
    n_val = int(0.75 * n)
    n_test = int(0.20 * n)
    n_play = n - n_val - n_test   # ensures no rounding problems

    df_val = df_shuffled.iloc[:n_val]
    df_test = df_shuffled.iloc[n_val:n_val + n_test]
    df_play = df_shuffled.iloc[n_val + n_test:]

    return df_val, df_test, df_play
```
