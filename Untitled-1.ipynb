```python
import time
import requests
import pandas as pd
import json 
import numpy as np

url = 'https://api.gdeltproject.org/api/v2/doc/doc' 
params = {
'query': '("wildlife crime" OR "poaching" OR "illegal fishing" OR "wildlife trade")',
'mode': 'artlist',
'maxrecords': 250,
'STARTDATETIME': 20251001000000,
'format' : 'json'
}
response = requests.get(url, params=params, timeout=30)

if response.status_code == 200 and "application/json" in response.headers.get("Content-Type", ""):
    payload = response.json()

    if "articles" in payload and len(payload["articles"]) > 0:
        df = pd.DataFrame(payload["articles"])
    else:
        df = pd.DataFrame()  # No results
else:
    raise RuntimeError(f"API failed: {response.status_code}")
```

```python
df["date"] = pd.to_datetime(df["seendate"], errors="coerce")
```
