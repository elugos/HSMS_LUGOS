```python
global_db_with_clusters = pd.read_csv('/data/elugos/global_db_with_clusters.csv')
```

```python
for cluster_id, group in global_db_with_clusters.groupby('cluster'):
    print(f"\n🟣 Cluster {cluster_id} ({len(group)} events)")
    print(group['EventCode'].value_counts())
    print(group['state'].value_counts())
    print()
```
