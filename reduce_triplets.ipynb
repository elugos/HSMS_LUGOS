```python


# optional: use nltk's stopwords if available
try:
    from nltk.corpus import stopwords
    STOPWORDS = set(stopwords.words("english"))
except Exception:
    # fallback minimal set
    STOPWORDS = {
        "the", "and", "of", "in", "on", "at", "a", "an", "for", "to",
        "is", "are", "was", "were", "be", "been", "by", "from", "with",
        "as", "that", "this", "it", "its", "or", "but", "not", "so"
    }


def filter_triples_by_title(globaldb: pd.DataFrame, ex_tr: pd.DataFrame) -> pd.DataFrame:
    """
    Keep rows in ex_tr where:
      - GlobalEventID exists in globaldb
      - At least one non-stopword token from globaldb.title
        appears in subject, relation, or object
    """
    # Merge only matching GlobalEventIDs
    merged = ex_tr.merge(globaldb[["GlobalEventID", "title"]], on="GlobalEventID", how="inner")

    # Tokenize & remove stopwords
    def tokenize_clean(text: str):
        tokens = re.findall(r"\b\w+\b", str(text).lower())
        return [t for t in tokens if t not in STOPWORDS and len(t) > 1]

    merged["title_tokens"] = merged["title"].apply(tokenize_clean)

    # Combine subject, relation, and object into one lowercase string
    merged["triple_text"] = (
        merged[["subject", "relation", "object"]]
        .astype(str)
        .agg(" ".join, axis=1)
        .str.lower()
    )

    # Keep only rows where any title token is in triple_text
    def token_in_triple(row):
        return any(tok in row["triple_text"] for tok in row["title_tokens"])

    filtered = merged[merged.apply(token_in_triple, axis=1)].copy()

    # Clean output
    filtered = filtered.drop(columns=["title_tokens", "triple_text"])
    return filtered
```

```python

def build_texts(df: pd.DataFrame, cols: List[str], joiner: str = " ") -> pd.Series:
    parts = [df[c].fillna("").astype(str).str.strip() if c in df.columns
             else pd.Series([""] * len(df), index=df.index) for c in cols]
    s = parts[0]
    for p in parts[1:]:
        s = s + joiner + p
    j = joiner.strip()
    if j:
        s = s.str.replace(rf"(?:\s*{j}\s*)+", f" {j} ", regex=True)
    return s.str.strip()


def encode_dataframe(
    df: pd.DataFrame,
    model: SentenceTransformer,
    cols: List[str] = ["subject", "relation", "object"],
    joiner: str = " ",
    batch_size: int = 256,
    normalize: bool = True,
) -> pd.DataFrame:
    """
    Encode specified columns from a DataFrame using a Sentence-BERT model.

    Args:
        df: Input DataFrame.
        model: Preloaded SentenceTransformer model.
        cols: Columns to join as text input.
        joiner: String used to join text columns.
        batch_size: Batch size for encoding.
        normalize: Whether to L2-normalize embeddings.

    Returns:
        DataFrame with added columns 'sbert_text' and 'embedding_json'.
    """
    # Warn about missing columns
    missing_cols = [c for c in cols if c not in df.columns]
    if missing_cols:
        print(f"[WARN] Missing columns {missing_cols}; treating them as empty strings.")

    sbert_text = build_texts(df, cols, joiner)
    embeddings = model.encode(
        sbert_text.tolist(),
        batch_size=batch_size,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=normalize,
    )

    out_df = df.copy()
    out_df["sbert_text"] = sbert_text
    out_df["embedding_json"] = [json.dumps(vec.tolist()) for vec in embeddings]
    return out_df
```

```python
def run_dbscan(df, eps=0.15, min_samples=5, scale=False, metric='cosine'):
    """
    Run DBSCAN clustering using vectors stored in 'embedding_json'.

    Args:
        df (pd.DataFrame): Must contain 'embedding_json' as a string like "[0.1, 0.2, ...]".
        eps (float): DBSCAN eps parameter (distance threshold).
        min_samples (int): DBSCAN min_samples parameter.
        scale (bool): Whether to standardize features before clustering.
        metric (str): Distance metric (e.g., 'cosine', 'euclidean').

    Returns:
        X_clustered (ndarray): Embedding matrix for non-noise rows only.
        clustered_df (pd.DataFrame): Rows with cluster labels >= 0 (noise removed).
        noise_df (pd.DataFrame): Rows with cluster == -1.
        summary (dict): {'n_clusters': int, 'n_noise': int}
    """
    df_work = df.copy(deep=True)

    # Parse embeddings
    s = df_work['embedding_json'].astype(str).str.strip()
    s = s.str.lstrip('[').str.rstrip(']')
    s = s.str.split(',')
    vecs = [np.array(list(map(float, parts)), dtype=np.float32) for parts in s]
    X = np.vstack(vecs)

    # Optional scaling
    if scale:
        X = StandardScaler().fit_transform(X)

    # Fit DBSCAN
    db = DBSCAN(eps=eps, min_samples=min_samples, metric=metric)
    labels = db.fit_predict(X)
    df_work['cluster'] = labels

    # Masks
    noise_mask = labels == -1
    cluster_mask = ~noise_mask

    # Split dfs
    noise_df = df_work.loc[noise_mask].drop(columns=['embedding_json']).reset_index(drop=True)
    clustered_df = df_work.loc[cluster_mask].drop(columns=['embedding_json']).reset_index(drop=True)

    # Extract only clustered embeddings
    X_clustered = X[cluster_mask]

    # Summary
    n_clusters = len(set(labels) - {-1})
    n_noise = int(noise_mask.sum())
    summary = {"n_clusters": n_clusters, "n_noise": n_noise}

    return X_clustered, clustered_df, noise_df
```

```python
def count_events_per_cluster(df: pd.DataFrame) -> pd.DataFrame:
    """
    Count how many times each GlobalEventID appears in each cluster.

    Args:
        df: DataFrame with columns ['GlobalEventID', 'cluster'].

    Returns:
        DataFrame with columns ['cluster', 'GlobalEventID', 'count'].
    """
    counts = (
        df.groupby(['cluster', 'GlobalEventID'])
          .size()
          .reset_index(name='count')
          .sort_values(['cluster', 'count'], ascending=[True, False])
    )
    return counts
```

```python
def build_event_graph(df: pd.DataFrame, df_info, min_count: int = 1) -> nx.Graph:
    """
    Build a co-occurrence graph from ['cluster','GlobalEventID','count'].
    - Node attrs:
        total_count: sum of 'count' across clusters
        clusters: set of clusters the ID appears in
    - Edge attrs:
        weight: number of clusters the pair co-occurred in
        clusters: set of clusters where the pair co-occurred
    """
    G = nx.Graph()

    for cluster_id, group in df.groupby("cluster"):
        # Map event -> count within this cluster
        counts = dict(zip(group["GlobalEventID"], group["count"]))
        events = list(counts.keys())


        # Add/update nodes
        for ev, c in counts.items():
            if ev not in G:
                G.add_node(ev, total_count=0, clusters=set(),GlobalEventID=ev)
            G.nodes[ev]["total_count"] += int(c)
            G.nodes[ev]["clusters"].add(cluster_id)

        # Add/update edges for co-occurring pairs within this cluster
        for u, v in itertools.combinations(events, 2):
            if G.has_edge(u, v):
                G[u][v]["weight"] += 1
                G[u][v]["clusters"].add(cluster_id)
            else:
                G.add_edge(u, v, weight=1, clusters={cluster_id})

    # Convert sets to sorted lists for cleaner serialization/hover text
    for n, data in G.nodes(data=True):
        data["clusters"] = sorted(data["clusters"])
    for u, v, data in G.edges(data=True):
        data["clusters"] = sorted(data["clusters"])
    

    # Build lookup dictionary from DataFrame
    info_dict = (
    df_info.drop_duplicates(subset="GlobalEventID", keep="first")
           .set_index("GlobalEventID")
           .to_dict(orient="index")
)

    # Enrich graph nodes if ID matches  
    #for node in G.nodes:
    #    if node in info_dict:
    #        nx.set_node_attributes(G, {node: info_dict[node]})

        # Build lookup dictionary from DataFrame

    # Get nodes that exist in both the graph and the dataframe
    valid_nodes = [n for n in G.nodes if n in info_dict]

    # Remove all nodes not in the DataFrame
    to_remove = set(G.nodes) - set(valid_nodes)
    G.remove_nodes_from(to_remove)

    # Now enrich only the remaining (matching) nodes
    nx.set_node_attributes(G, {n: info_dict[n] for n in valid_nodes})

            
        # 5) Remove edges that don’t meet min_count
    edges_to_remove = [(u, v) for u, v, d in G.edges(data=True) if d["weight"] < min_count]
    G.remove_edges_from(edges_to_remove)

    return G
```

```python
def create_graph_traces(G, pos_attribute_name):
    """
    Creates the node trace and edge trace for input graph `G` with nodes having attribute `pos_attribute_name`.

    Args:
        G (nx.Graph) : The graph.
        pos_attribute_name (str) : The attribute name for the node position.
    
    Returns:
        node_trace, edge_trace
    """
    edge_x = []
    edge_y = []
    for edge in G.edges():
        x0, y0 = G.nodes[edge[0]][pos_attribute_name]
        x1, y1 = G.nodes[edge[1]][pos_attribute_name]
        edge_x.append(x0)
        edge_x.append(x1)
        edge_x.append(None)
        edge_y.append(y0)
        edge_y.append(y1)
        edge_y.append(None)

    edge_trace = go.Scatter(
        x=edge_x, y=edge_y,
        line=dict(width=0.5, color='#888'),
        hoverinfo='none',
        mode='lines')

    node_x = []
    node_y = []
    for node in G.nodes():
        x, y = G.nodes[node][pos_attribute_name]
        node_x.append(x)
        node_y.append(y)

    node_trace = go.Scatter(
        x=node_x, y=node_y,
        mode='markers',
        hoverinfo='text',
        marker=dict(
            showscale=True,
            # colorscale options
            #'Greys' | 'YlGnBu' | 'Greens' | 'YlOrRd' | 'Bluered' | 'RdBu' |
            #'Reds' | 'Blues' | 'Picnic' | 'Rainbow' | 'Portland' | 'Jet' |
            #'Hot' | 'Blackbody' | 'Earth' | 'Electric' | 'Viridis' |
            colorscale='YlGnBu_r',
            reversescale=True,
            color=[],
            size=10,
            colorbar=dict(
                thickness=15,
                title=dict(
                text='Node Connections',
                side='right'
                ),
                xanchor='left',
            ),
            line_width=2))

    return node_trace, edge_trace
```

```python
def color_nodes_by_state(G, node_trace):
    """
    Color Plotly node trace by the 'state' attribute in the graph.

    Each unique state gets a distinct color index.
    If a node has no state, it will be assigned a default color (e.g., gray).

    Args:
        G: networkx.Graph with 'state' attributes on nodes.
        node_trace: Plotly Scatter trace (from network visualization).
    """

    # Extract state attribute for each node
    node_states = nx.get_node_attributes(G, 'state')

    # Build sorted list of unique states
    unique_states = sorted(set(node_states.values()))

    # Assign each state a numeric color index
    color_map = {state: i for i, state in enumerate(unique_states)}

    # Normalize colors to 0–1 range for Plotly colormap
    norm = mcolors.Normalize(vmin=0, vmax=len(unique_states) - 1)
    cmap = cm.get_cmap('tab20', len(unique_states))  # you can pick 'viridis', 'plasma', etc.

    node_colors = []
    node_text = []

    for node in G.nodes():
        state = node_states.get(node, None)
        if state is not None:
            color_val = cmap(norm(color_map[state]))
            node_colors.append(mcolors.to_hex(color_val))
            node_text.append(f"State: {state}")
        else:
            node_colors.append("#CCCCCC")  # gray for missing
            node_text.append("State: None")

    node_trace.marker.color = node_colors
    node_trace.text = node_text
```

```python
def create_network_graph_state(G, node_trace, edge_trace,
                         title='',
                         annotation_text=''):
    """
    Create a Plotly network graph showing:
      - GlobalEventID, EventCode, and Title on hover
      - Nodes colored by 'state' with legend
      - No edges shown
    """
    # --- gather node info ---
    node_states = nx.get_node_attributes(G, 'state')
    unique_states = sorted(set(node_states.values()))

    # color palette
    cmap = cm.get_cmap('tab20', len(unique_states))
    color_map = {s: mcolors.to_hex(cmap(i)) for i, s in enumerate(unique_states)}

    # --- build hover text and assign colors ---
    xs, ys, colors, texts, states = [], [], [], [], []
    for node in G.nodes():
        data = G.nodes[node]
        x, y = data.get("pos_sp", [None, None])
        xs.append(x)
        ys.append(y)

        geid = data.get("GlobalEventID", node)
        code = data.get("EventCode", "N/A")
        title_str = data.get("title", "N/A")
        state = data.get("state", "N/A")

        texts.append(
            f"<b>GlobalEventID:</b> {geid}<br>"
            f"<b>EventCode:</b> {code}<br>"
            f"<b>Title:</b> {title_str}"
        )
        colors.append(color_map.get(state, "#CCCCCC"))
        states.append(state)

    # --- create one trace per state (for legend) ---
    node_traces = []
    for state in unique_states:
        indices = [i for i, s in enumerate(states) if s == state]
        if not indices:
            continue
        trace = go.Scatter(
            x=[xs[i] for i in indices],
            y=[ys[i] for i in indices],
            mode='markers',
            name=state,  # legend label
            marker=dict(size=8, color=color_map[state], line_width=0.5),
            text=[texts[i] for i in indices],
            hoverinfo='text'
        )
        node_traces.append(trace)

    # --- build figure ---
    fig = go.Figure(
        data=node_traces,  # omit edge_trace for no connections
        layout=go.Layout(
            title=dict(text=title, font=dict(size=16)),
            showlegend=True,
            hovermode='closest',
            margin=dict(b=20, l=5, r=5, t=40),
            annotations=[
                dict(
                    text=annotation_text,
                    showarrow=False,
                    xref="paper", yref="paper",
                    x=0.005, y=-0.002,
                )
            ],
            xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        ),
    )

    fig.show()
```

```python
def create_network_graph(
    G,
    node_trace,
    edge_trace,
    title='',
    annotation_text='',
    cameo_codes=None
):
    """
    Create a Plotly network graph showing:
      - GlobalEventID, EventCode, Title, and State on hover
      - Nodes colored by CAMEO event type (from EventCode)
      - Legend shows CAMEO code + description
      - Edge trace optional (can be empty)
    """
    if cameo_codes is None:
        raise ValueError("You must provide the cameo_codes dictionary.")

    # invert cameo dict for numeric→description lookup
    cameo_lookup = {v: k for k, v in cameo_codes.items()}

    # Extract numeric event code endings
    node_codes = {}
    for n, data in G.nodes(data=True):
        code = str(data.get("EventCode", ""))
        match = re.search(r"(\d+)$", code)
        node_codes[n] = match.group(1) if match else None

    unique_codes = sorted(
        {c for c in node_codes.values() if c is not None}, key=lambda x: int(x)
    )
    cmap = cm.get_cmap("tab20", len(unique_codes))
    color_map = {code: mcolors.to_hex(cmap(i)) for i, code in enumerate(unique_codes)}

    # --- Gather node data ---
    xs, ys, texts, colors, codes = [], [], [], [], []
    for node in G.nodes():
        data = G.nodes[node]
        x, y = data.get("pos_fa2", [None, None])  # use same layout key
        xs.append(x)
        ys.append(y)

        geid = data.get("GlobalEventID", node)
        event_code = data.get("EventCode", "N/A")
        state = data.get("state", "N/A")
        title_str = data.get("title", "N/A")

        cameo_num = node_codes.get(node)
        cameo_desc = cameo_lookup.get(cameo_num, "Unknown")

        colors.append(color_map.get(cameo_num, "#CCCCCC"))
        codes.append(cameo_num)
        texts.append(
            f"<b>GlobalEventID:</b> {geid}<br>"
            f"<b>EventCode:</b> {event_code}<br>"
            f"<b>CAMEO:</b> {cameo_num}: {cameo_desc}<br>"
            f"<b>State:</b> {state}<br>"
            f"<b>Title:</b> {title_str}"
        )

    # --- Build one trace per CAMEO code for color legend ---
    node_traces = []
    for code in unique_codes:
        idxs = [i for i, c in enumerate(codes) if c == code]
        if not idxs:
            continue
        trace = go.Scatter(
            x=[xs[i] for i in idxs],
            y=[ys[i] for i in idxs],
            mode="markers",
            name=f"{code}: {cameo_lookup.get(code, 'Unknown')}",
            marker=dict(size=8, color=color_map[code], line_width=0.5),
            text=[texts[i] for i in idxs],
            hoverinfo="text",
        )
        node_traces.append(trace)

    # --- Combine with existing edge trace (if given) ---
    #traces = []
    #if edge_trace is not None:
    #    traces.append(edge_trace)
    #traces.extend(node_traces)

    # --- Build figure ---
    fig = go.Figure(
        data=node_traces,
        layout=go.Layout(
            title=dict(text=title, font=dict(size=16)),
            showlegend=True,
            hovermode="closest",
            margin=dict(b=20, l=5, r=5, t=40),
            annotations=[
                dict(
                    text=annotation_text,
                    showarrow=False,
                    xref="paper", yref="paper",
                    x=0.005, y=-0.002,
                )
            ],
            xaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
            yaxis=dict(showgrid=False, zeroline=False, showticklabels=False),
        ),
    )

    fig.show()
```

```python
def spectral_embed_from_adjacency(
    A: np.ndarray,
    n_components: int = 2,
    node_ids: Optional[Union[np.ndarray, list]] = None,
    drop_isolates: bool = False,
    make_symmetric: bool = False,
    zero_self_loops: bool = False,
    use_sparse: bool = False,
    **kwargs
) -> pd.DataFrame:
    """
    Compute a spectral embedding from an adjacency matrix and return a DataFrame
    with columns ['x','y', ...]. Works with dense or sparse input.

    Args:
        A : np.ndarray (n x n)
            Adjacency (non-negative). If not perfectly symmetric, set make_symmetric=True.
        n_components : int, default 2
            Dimensions of the embedding.
        node_ids : array-like of shape (n,), optional
            Labels to associate with rows. Defaults to range(n).
        drop_isolates : bool, default False
            If True, nodes with degree 0 are dropped from the output.
            If False, isolates are kept with NaN coordinates.
        make_symmetric : bool, default True
            If True, symmetrize A as (A + A.T) / 2.
        zero_self_loops : bool, default True
            If True, zero out the diagonal of A.
        use_sparse : bool, default True
            If True, pass a CSR matrix to sklearn (recommended for large graphs).
        **kwargs :
            Extra args passed to sklearn.manifold.SpectralEmbedding, e.g.:
            - eigen_solver={'arpack','amg','lobpcg'}
            - random_state=42
            Note: We set affinity='precomputed' internally.

    Returns:
        pd.DataFrame:
            Columns: ['node_id','x','y', ...] (or just coords if node_ids not provided).
            If drop_isolates=False, isolates have NaN for coordinates.
    """
    A = np.asarray(A)
    if A.ndim != 2 or A.shape[0] != A.shape[1]:
        raise ValueError("A must be a square (n x n) adjacency matrix.")

    n = A.shape[0]
    if node_ids is None:
        node_ids = np.arange(n)
    else:
        node_ids = np.asarray(node_ids)
        if node_ids.shape[0] != n:
            raise ValueError("node_ids must have length equal to A.shape[0].")

    # Basic hygiene on adjacency
    if make_symmetric:
        A = 0.5 * (A + A.T)
    if zero_self_loops:
        np.fill_diagonal(A, 0.0)

    # Identify isolates (degree 0)
    degrees = A.sum(axis=1)
    valid_mask = degrees > 0

    # Build the submatrix for non-isolates
    if valid_mask.any():
        A_valid = A[np.ix_(valid_mask, valid_mask)]
        if use_sparse:
            A_valid = sparse.csr_matrix(A_valid)

        # SpectralEmbedding expects an affinity matrix; adjacency is OK when using 'precomputed'
        emb = SpectralEmbedding(
            n_components=n_components,
            affinity="precomputed",
            **kwargs
        )
        coords_valid = emb.fit_transform(A_valid)  # (n_valid, n_components)
    else:
        # All isolates; produce an all-NaN coordinate matrix
        coords_valid = np.empty((0, n_components))

    # Build column names: x, y, x2, x3, ...
    col_names = []
    if n_components >= 1: col_names.append("x")
    if n_components >= 2: col_names.append("y")
    for i in range(2, n_components):
        col_names.append(f"x{i}")

    # Prepare output with NaNs, then fill for non-isolates
    out_coords = np.full((n, n_components), np.nan, dtype=float)
    out_coords[valid_mask, :] = coords_valid

    # Assemble DataFrame
    df_out = pd.DataFrame(out_coords, columns=col_names)
    df_out.insert(0, "node_id", node_ids)

    if drop_isolates:
        df_out = df_out.loc[valid_mask].reset_index(drop=True)

    return df_out
```

```python
def run_dbscan_on_graph(G, pos_key='pos_sp', eps=0.15, min_samples=5, metric='cosine'):
    """
    Run DBSCAN on node positions or embeddings stored under a node attribute.

    Args:
        G : nx.Graph
        pos_key : str
            Node attribute containing coordinates (e.g. 'pos_sp' or 'embedding').
        eps : float
            DBSCAN eps parameter.
        min_samples : int
            Minimum number of points for a cluster.
        metric : str
            Distance metric ('euclidean', 'cosine', etc.).

    Returns:
        labels : np.ndarray of shape (n_nodes,)
    """
    # Extract coordinate vectors
    coords = []
    nodes = list(G.nodes())
    for n in nodes:
        v = G.nodes[n].get(pos_key)
        if v is None:
            raise ValueError(f"Node {n} missing '{pos_key}' attribute.")
        coords.append(v)
    X = np.array(coords, dtype=float)

    # Run DBSCAN
    db = DBSCAN(eps=eps, min_samples=min_samples, metric=metric).fit(X)
    labels = db.labels_

    # Store cluster labels
    nx.set_node_attributes(G, {n: int(l) for n, l in zip(nodes, labels)}, 'cluster')

    print(f"DBSCAN found {len(set(labels)) - (1 if -1 in labels else 0)} clusters "
          f"and {np.sum(labels == -1)} noise points.")
    return labels
```

```python
def run_dbscan_on_node_embeddings(
    df: pd.DataFrame,
    id_col: str = 'node_id',
    eps: float = 0.3,
    min_samples: int = 5,
    scale: bool = True,
    metric: str = 'euclidean'
):
    """
    Run DBSCAN clustering on all numeric columns except the node ID column.

    Args:
        df : pd.DataFrame
            Must contain 'node_id' and numeric embedding columns (e.g. x, y, x2, ...).
        id_col : str
            Name of the node ID column to exclude.
        eps : float
            DBSCAN distance threshold (tune carefully for embeddings).
        min_samples : int
            Minimum number of samples to form a cluster.
        scale : bool
            Whether to standardize features before clustering.
        metric : str
            Distance metric ('cosine' is common for embeddings).

    Returns:
        df_out : pd.DataFrame
            Copy of df with a new 'cluster' column.
        summary : dict
            {'n_clusters': int, 'n_noise': int}
    """
    df_out = df.copy(deep=True)

    # Select all numeric columns except the ID
    feature_cols = [c for c in df_out.columns if c != id_col and np.issubdtype(df_out[c].dtype, np.number)]
    if not feature_cols:
        raise ValueError("No numeric feature columns found for clustering.")

    X = df_out[feature_cols].to_numpy(dtype=float)

    if scale:
        X = StandardScaler().fit_transform(X)

    db = DBSCAN(eps=eps, min_samples=min_samples, metric=metric).fit(X)
    df_out['cluster'] = db.labels_

    summary = {
        'n_clusters': len(set(db.labels_)) - (1 if -1 in db.labels_ else 0),
        'n_noise': int(np.sum(db.labels_ == -1))
    }

    print(f"DBSCAN found {summary['n_clusters']} clusters and {summary['n_noise']} noise points.")
    return df_out, summary
```

```python
def reduce_embeddings_to_tsne(
    df: pd.DataFrame,
    id_col: str = 'node_id',
    n_components: int = 2,
    perplexity: float = 30.0,
    learning_rate: float = 200.0,
    max_iter: int = 1000,
    scale: bool = True,
    random_state: int = 42
) -> pd.DataFrame:
    """
    Reduce high-dimensional embeddings to 2D using t-SNE.

    Args:
        df : pd.DataFrame
            Must contain numeric embedding columns.
        id_col : str
            Column identifying each row/node.
        n_components : int
            Output dimensions (default 2).
        perplexity : float
            Balances local/global structure (typical: 5–50).
        learning_rate : float
            Step size for optimization.
        n_iter : int
            Number of iterations.
        scale : bool
            Standardize before t-SNE.
        random_state : int
            Reproducibility seed.

    Returns:
        df_out : pd.DataFrame
            Copy of df with 'dim1' and 'dim2' columns added.
    """
    # Select numeric embedding columns (exclude id + cluster)
    feature_cols = [
        c for c in df.columns
        if c not in [id_col, 'cluster'] and np.issubdtype(df[c].dtype, np.number)
    ]
    X = df[feature_cols].to_numpy(dtype=float)

    # Handle missing values if any
    if np.isnan(X).any():
        print("⚠️ NaN values found — filling with column means before t-SNE.")
        X = np.where(np.isnan(X), np.nanmean(X, axis=0), X)

    if scale:
        X = StandardScaler().fit_transform(X)

    tsne = TSNE(
        n_components=n_components,
        perplexity=perplexity,
        learning_rate=learning_rate,
        max_iter=max_iter,
        random_state=random_state,
        init='pca',
        n_jobs=-1  # parallel if available
    )
    reduced = tsne.fit_transform(X)

    df_out = df.copy(deep=True)
    df_out['dim1'] = reduced[:, 0]
    df_out['dim2'] = reduced[:, 1]
    print(f"✅ t-SNE reduction complete ({len(df_out)} points → {n_components}D)")
    return df_out
```

```python
def join_global_db_with_clusters(
    df_clustered: pd.DataFrame,
    global_db: pd.DataFrame,
    node_col: str = 'node_id',
    global_id_col: str = 'GlobalEventID'
) -> pd.DataFrame:
    """
    Join cluster labels and 2D reduced coordinates (dim1, dim2)
    from df_clustered into global_db, keeping only rows that exist
    in df_clustered.
    """
    required_cols = [node_col, 'cluster', 'dim1', 'dim2']
    missing = [c for c in required_cols if c not in df_clustered.columns]
    if missing:
        raise ValueError(f"Missing required columns in df_clustered: {missing}")

    merged = global_db.merge(
        df_clustered[required_cols],
        left_on=global_id_col,
        right_on=node_col,
        how='inner'  # only keep matching rows
    )

    merged.drop(columns=[node_col], inplace=True)

    print(f"✅ Kept {len(merged)} rows (out of {len(global_db)}).")
    print(f"Clusters found: {merged['cluster'].nunique()} (including noise if present).")
    print(f"dim1/dim2 attached for visualization.")

    return merged
```
