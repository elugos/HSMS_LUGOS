```python
from logistic import prepare_dataset, train_tfidf_logistic, LABELS
from sklearn.model_selection import train_test_split
import pandas as pd
```

```python
input_file = "../data/train_20251216161539.csv"
df = pd.read_csv(input_file)
```

```python
X_text, Y = prepare_dataset(df)
print(f"Data points: {len(X_text)}")
X_train, X_test, Y_train, Y_test = train_test_split(X_text, Y, 
                                                    test_size=0.2, random_state=7,
                                                    # stratify=[tuple(y) for y in Y]
                                                    )

X_train, X_valid, Y_train, Y_valid = train_test_split(X_train, Y_train,
                                                      test_size=0.2, random_state=3)
```

```python
clf = train_tfidf_logistic(X_train, Y_train)
```

```python
from logistic import tune_thresholds_f1
# Tune probability thresholds on the validation set
Y_pred_valid = clf.predict_proba(X_valid)
p_class = tune_thresholds_f1(Y_valid, Y_pred_valid)
```

```python
from sklearn.metrics import classification_report

Y_pred = clf.predict_proba(X_test)
Y_hat = (Y_pred >= p_class).astype(int)

print(classification_report(Y_test, Y_hat, target_names=LABELS))
```
