```python
def gdelt_db(df):
    new_df = pd.DataFrame({
    "GlobalEventID": df["GlobalEventID"],
    "date": df["date"],           
    "EventCode": df[["EventCode", "EventBaseCode", "EventRootCode"]]
        .astype(str)
        .agg("_".join, axis=1),
    "state": df["ActionGeo_FeatureID"],
    "title" : df["title"]})

    return new_df

```
